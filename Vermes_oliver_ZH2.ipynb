```sql
Select distinct sz.szoba_id,
        sz.ferohely,
        Count(*) Over(partition By sz.szoba_id) As 'foglalasok_szama',
        Dense_rank() Over(Order By sz.ferohely ) As 'rangsor'
From Foglalas f Join Szoba sz On sz.SZOBA_ID=f.SZOBA_FK
Order By 3 desc, 2 asc
```

```sql
SELECT f.*
From foglalas f join Vendeg v On f.ugyfel_fk=v.USERNEV
WHERE v.szul_dat=
(
    Select Max(szul_dat)
    From Vendeg
    Where SZUL_DAT<
    (
        Select Max(szul_dat)
        From Vendeg
    )
)
```
